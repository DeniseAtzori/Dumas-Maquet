```python
import re
import os

def pulisci_testi(cartella_input, cartella_output):
    # Crea la cartella di output se non esiste
    if not os.path.exists(cartella_output):
        os.makedirs(cartella_output)

    # Regex per trovare i singoli ritorni a capo (non preceduti né seguiti da altri \n)
    # Gestisce sia i formati Linux/Mac (\n) che Windows (\r\n)
    regex_ritorno_singolo = re.compile(r'(?<!\n)\n(?!\n)')

    for nome_file in os.listdir(cartella_input):
        if nome_file.endswith(".txt"):
            percorso_completo = os.path.join(cartella_input, nome_file)
            
            with open(percorso_completo, 'r', encoding='utf-8') as f:
                contenuto = f.read()

            # 1. Normalizziamo i ritorni a capo di Windows (\r\n) in (\n)
            contenuto = contenuto.replace('\r\n', '\n')

            # 2. Applichiamo la Regex per togliere i ritorni a capo nei paragrafi
            pulito = regex_ritorno_singolo.sub(' ', contenuto)

            # 3. Opzionale: Rimuoviamo eventuali spazi doppi creati dalla sostituzione
            pulito = re.sub(r' +', ' ', pulito)

            # Salviamo il risultato nel nuovo file
            percorso_salvataggio = os.path.join(cartella_output, nome_file)
            with open(percorso_salvataggio, 'w', encoding='utf-8') as f:
                f.write(pulito)
            
            print(f"Processato: {nome_file}")

# --- CONFIGURAZIONE ---
# Inserisci qui il nome della cartella dove si trovano i tuoi file
cartella_sorgente = "testi_prova" 
cartella_destinazione = "testi_puliti"

pulisci_testi(cartella_sorgente, cartella_destinazione)
print("\nOperazione completata!")
```
