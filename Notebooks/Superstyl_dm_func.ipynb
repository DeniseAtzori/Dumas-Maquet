```python
import superstyl as sty
import functionwordsets as fw
```

```python
fr = fw.load() 
#list(fr.all)
fw = [[feat.strip(), 0] for feat in fr.all]
```

```python
import glob
path_txt = glob.glob("Dumas-Maquet/Corpus/Txt_nettoyes/*")
path_txt
```

```python
config = sty.Config.from_dict(
    {
  "corpus": 
    {
       "paths": path_txt,
       "format": "txt",
    },
  "features": [
    {
      "name": "fw",
      "type": "words",
      "n": 1,
      "k": 1000,
      "freq_type": "relative",
      "culling": 0.1,
      "feat_list": fw, 
    },
  ],
  "normalization": {
      "keep_punct": False,
      "keep_sym": False,
      "no_ascii": False
  },     
  "sampling": {
    "enabled": True,
    "units": "words",
    "sample_size": 3000,
    #"sample_step": 500,
    "max_samples": 20,
    "samples_random": True
  },
  "svm": {
    "cross_validate": "group-k-fold",
    "norms": True,
    "balance": "downsampling",
    "class_weights": True,
    "final_pred": False,
    "get_coefs": True
  }       
} 
)
```

```python
corpus, feats = sty.load_corpus(config)
```
